```python
# Reading csv file from colab uploaded files to pandas Data Frame
import pandas as pd
dataset_old=pd.read_csv('/content/census_2011 - census_2011.csv.csv')
dataset_old
```

```python
# Renaming some column names for uniformity
dataset=dataset_old.rename(columns={'State name':'State/UT','District name':'District','Male_Literate':'Literate_Male','Female_Literate':'Literate_Female','Rural_Households':'Households_Rural','Urban_Households':'Households_Urban','Age_Group_0_29':'Young_and_Adult','Age_Group_30_49':'Middle_Aged','Age_Group_50':'Senior_Citizen','Age not stated':'Age_Not_Stated','District code':'District_Code','Young and_Adult':'Young_and_Adult'})
```

```python
# Replacing the first character of State/UT to upper case and the rest to lower case, using 'title' for uniformity
dataset['State/UT']=dataset['State/UT'].apply(str.title)
```

```python
# Replacing the names for uniformity
dataset=dataset.replace(to_replace=['Andaman And Nicobar Islands','Jammu And Kashmir','Nct Of Delhi','Daman And Diu','Dadra And Nagar Haveli'],value=['Andaman and Nicobar Islands','Jammu and Kashmir','Nct of Delhi','Daman and Diu','Dadra and Nagar Haveli'])
```

```python
# State name for some Districts, are changed to Telangana after state division from Andhra Pradesh
District_Telangana=['Adilabad','Nizamabad','Karimnagar','Medak','Hyderabad','Rangareddy','Mahbubnagar','Nalgonda','Warangal','Khammam']

dataset.loc[dataset['District'].isin(District_Telangana),'State/UT']='Telangana'
```

```python
# State name for some Districts, are changed to Ladakh after state division from Jammu and Kashmir
District_Ladakh=['Leh(Ladakh)','Kargil']

dataset.loc[dataset['District'].isin(District_Ladakh),'State/UT']='Ladakh'
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import sqlite3
from sqlalchemy import create_engine,text
path='sqlite:///census.db'
my_conn=create_engine(path).connect()

st.write(''' The task is to clean, process, and analyze census data from a given source.The goal is to ensure uniformity,
             accuracy, and accessibility of the census data for further analysis and visualization.''')


##1. What is the total population of each district?

df=pd.read_sql(text('select District,Population from Census_2011'), my_conn)
st.write('Total population of each district')
st.write(df)

#2. How many literate males and females are there in each district?

df=pd.read_sql(text('select District,Literate_Male,Literate_Female from Census_2011'), my_conn)
st.write('Literate males and females in each district')
st.write(df)

#3. What is the percentage of workers (both male and female) in each district?

df=pd.read_sql(text('''select District,Workers,Male_Workers,Female_Workers,
(Male_Workers/Workers)*100 as Male_Percent, (Female_Workers/Workers)*100 as Female_Percent
from Census_2011'''), my_conn)
st.write('Percentage of workers (both male and female) in each district')
st.write(df)

#4.How many households have access to LPG or PNG as a cooking fuel in each district?

df=pd.read_sql(text('select District,LPG_or_PNG_Households from Census_2011'), my_conn)
st.write('Households having access to LPG or PNG as a cooking fuel in each district')
st.write(df)

#5. What is the religious composition (Hindus, Muslims, Christians, etc.) of each district?

df=pd.read_sql(text('''select District,(Hindus/Population)*100 as Hindus_Perc,
(Muslims/Population)*100 as Muslims_Perc,
(Christians/Population)*100 as Christians_Perc,
(Sikhs/Population)*100 as Sikhs_Perc,
(Buddhists/Population)*100 as Buddhists_Perc,
(Jains/Population)*100 as Jains_Perc,
(Others_Religions/Population)*100 as Other_Religions_Perc,
(Religion_Not_Stated/Population)*100 as Religion_Not_Stated_Perc from Census_2011'''), my_conn)
st.write('Religious composition in each district')
st.write(df)

#6. How many households have internet access in each district?

df=pd.read_sql(text('select District,Households_with_Internet from Census_2011'), my_conn)
st.write('Households having internet access in each district')
st.write(df)

# 7.What is the educational attainment distribution (below primary, primary, middle, secondary, etc.)
# in each district?

df=pd.read_sql(text('''select District,Below_Primary_Education,Primary_Education,Middle_Education,
Secondary_Education,Higher_Education,Graduate_Education,Other_Education from Census_2011'''), my_conn)
st.write('Educational attainment distribution in each district')
st.write(df)

# 8. How many households have access to various modes of transportation
# (bicycle, car, radio, television, etc.) in each district?

df=pd.read_sql(text('''select District,Households_with_Bicycle,Households_with_Car_Jeep_Van,
Households_with_Scooter_Motorcycle_Moped,
Households_with_TV_Computer_Laptop_Telephone_mobile_phone_and_Scooter_Car from Census_2011'''),my_conn)
st.write('Households having access to various modes of transportation')
st.write(df)

# 9. What is the condition of occupied census houses (dilapidated,
#with separate kitchen, with bathing facility, with latrine facility, etc.) in each district?

df=pd.read_sql(text('''select District,Condition_of_occupied_census_houses_Dilapidated_Households,
Households_with_separate_kitchen_Cooking_inside_house,Having_bathing_facility_Total_Households,
Having_latrine_facility_within_the_premises_Total_Households from Census_2011'''),my_conn)
st.write('Condition of occupied census houses in each district')
st.write(df)

# 10.How is the household size distributed (1 person, 2 persons, 3-5 persons, etc.) in each district?

df=pd.read_sql(text('''select District,Household_size_1_person_Households,
Household_size_2_persons_Households,
Household_size_1_to_2_persons,
Household_size_3_persons_Households,
Household_size_3_to_5_persons_Households,
Household_size_4_persons_Households,
Household_size_5_persons_Households,
Household_size_6_8_persons_Households,
Household_size_9_persons_and_above_Households from Census_2011'''),my_conn)
st.write('Household size distribution in each district')
st.write(df)

# 11.What is the total number of households in each state?

df=pd.read_sql(text('select `State/UT`,sum(Households) from Census_2011 group by `State/UT`'),my_conn)
st.write('Total number of households in each state')
st.write(df)

# 12.How many households have a latrine facility within the premises in each state?

df=pd.read_sql(text('''select `State/UT`,
sum(Having_latrine_facility_within_the_premises_Total_Households)
from Census_2011 group by `State/UT`'''),my_conn)
st.write('Households having latrine facility within the premises in each state')
st.write(df)

# 13.What is the average household size in each state?

df=pd.read_sql(text('select `State/UT`,avg(Households) from Census_2011 group by `State/UT`'),my_conn)
st.write('Average household size in each state')
st.write(df)

# 14. How many households are owned versus rented in each state?

df=pd.read_sql(text('''select `State/UT`,sum(Ownership_Owned_Households),sum(Ownership_Rented_Households)
from Census_2011 group by `State/UT`'''),my_conn)
st.write('Households owned versus rented in each state')
st.write(df)

# 15.What is the distribution of different types of latrine facilities
 #(pit latrine, flush latrine, etc.) in each state?

df=pd.read_sql(text('''select `State/UT`,Type_of_latrine_facility_Pit_latrine_Households,
sum(Type_of_latrine_facility_Other_latrine_Households),
sum(Type_of_latrine_facility_Night_soil_disposed_into_open_drain_Households),
sum(Type_of_latrine_facility_Flush_pour_flush_latrine_connected_to_other_system_Households)
from census_2011 group by `State/UT`'''),my_conn)
st.write('Distribution of different types of latrine facilities in each state')
st.write(df)

# 16.How many households have access to drinking water sources near the premises in each state?

df=pd.read_sql(text('''select `State/UT`,sum(Location_of_drinking_water_source_Near_the_premises_Households)
 from Census_2011 group by `State/UT`'''),my_conn)
st.write('Households having access to drinking water sources near the premises in each state')
st.write(df)

# 17.What is the average household income distribution in each state based on the power parity categories?

df=pd.read_sql(text('''select `State/UT`,
avg(Power_Parity_Less_than_Rs_45000),
avg(Power_Parity_Rs_45000_90000),
avg(Power_Parity_Rs_90000_150000),
avg(Power_Parity_Rs_45000_150000),
avg(Power_Parity_Rs_150000_240000),
avg(Power_Parity_Rs_240000_330000),
avg(Power_Parity_Rs_150000_330000),
avg(Power_Parity_Rs_330000_425000),
avg(Power_Parity_Rs_425000_545000),
avg(Power_Parity_Rs_330000_545000),
avg(Power_Parity_Above_Rs_545000) from Census_2011 group by `State/UT`'''),my_conn)
st.write('Average household income distribution in each state based on the power parity categories')
st.write(df)

# 18. What is the percentage of married couples with different household sizes in each state?

df=pd.read_sql(text('''select `State/UT`,
sum((Married_couples_1_Households/(Married_couples_1_Households+Married_couples_2_Households+
Married_couples_3_or_more_Households+Married_couples_None_Households))*100),
sum((Married_couples_2_Households/(Married_couples_1_Households+Married_couples_2_Households+
Married_couples_3_or_more_Households+Married_couples_None_Households))*100),
sum((Married_couples_3_or_more_Households/(Married_couples_1_Households+Married_couples_2_Households+
Married_couples_3_or_more_Households+Married_couples_None_Households))*100),
sum((Married_couples_None_Households/(Married_couples_1_Households+Married_couples_2_Households+
Married_couples_3_or_more_Households+Married_couples_None_Households))*100) from Census_2011 group by `State/UT`'''),my_conn)
st.write('percentage of married couples with different household sizes in each state')
st.write(df)

# 19. How many households fall below the poverty line in each state based on the power parity categories?

df=pd.read_sql(text('select `State/UT`,sum(Power_Parity_Less_than_Rs_45000) from census_2011 group by `State/UT`'),my_conn)
st.write('Households below the poverty line in each state based on the power parity categories')
st.write(df)

# 20. What is the overall literacy rate (percentage of literate population) in each state?

df=pd.read_sql(text('select `State/UT`,sum((Literate/Population)*100) from Census_2011 group by `State/UT`'),my_conn)
st.write('Overall literacy rate in each state')
st.write(df)
```
